# Beyond the Price Tag - CRISP-DM Analysis of Premier League Player Performance and Market Value


## Data Analytics | CRISP-DM CA2 | Ahmad Shahzad


### 1. Introduction

The football transfer market is a competitive environment in which
clubs must make important financial decisions when recruiting players.
Identifying players who demonstrate strong on-pitch performance
relative to their estimated market value could provide clubs with
useful insights when making recruitment decisions.

This project will investigate the relationship between Premier
League player performance statistics and estimated player market
values. Using data analytics techniques, the aim is to explore
whether statistical performance can help identify players who may
represent potential value in the transfer market.

The project will follow the CRISP-DM (Cross-Industry Standard Process
for Data Mining) lifecycle to structure the analysis, from defining
the business problem through to evaluating the results and presenting
practical recommendations.

### 1.2 Project Aim

The aim of the project is to investigate whether Premier League player performance data can help identify players whose estimated market value appear relatively low compared with their statistical performance.

The analysis will explore the relationship between player characteristics, performance statistics and estimated market value, with the intention of supporting more informed and cost-conscious football recruitment decisions.

### 1.3 Main Research Question

Can football player performance statistics be used to identify potentially undervalued players whose on-pitch performances suggest they may be worth more than their current estimated market value

### 2.1 Project Plan

The project will be completed using the six CRISP-DM phases.
Each phase will contribute towards answering the main research
question.
| CRISP-DM Phase | Planned Activities |
|---|---|
| Business Understanding | Define the recruitment problem, research question, stakeholders and success criteria. |
| Data Understanding | Find suitable Premier League performance and market value datasets and investigate their quality. |
| Data Preparation | Clean the datasets, handle missing values, combine relevant information and prepare variables for analysis. |
| Modelling | Develop a baseline and suitable predictive model to estimate player market values. |
| Evaluation | Evaluate model performance and investigate players whose observed values differ from model predictions. |
| Deployment | Present findings, identify potential recruitment insights and discuss limitations and recommendations. |
